# Part 1 : XML 파싱 (p.9 ~ p.13)

##  p.9 — ElementTree로 XML 파일 읽기

**menu.xml 구조**
```xml
<menu>
  <breakfast hours="7-11">
    <item price="$6.00">breakfast burritos</item>
    <item price="$4.00">pancakes</item>
  </breakfast>
  <lunch hours="11-3">
    <item price="$5.00">hamburger</item>
  </lunch>
  <dinner hours="3-10">
    <item price="8.00">spaghetti</item>
  </dinner>
</menu>
```

````python
import xml.etree.ElementTree as et

# XML 파일 파싱 → tree 객체 생성
tree = et.ElementTree(file='menu.xml')

# 루트 엘리먼트 가져오기
root = tree.getroot()

# 루트 태그 이름 출력
print(root.tag)          # menu
````

````python
# root 의 타입 확인
type(root)               # xml.etree.ElementTree.Element
````

---
##  p.10 — 내용 프린팅 (child / grandchild 순회)

````python
# root(menu)의 자식(breakfast, lunch, dinner)과
# 손자(item) 태그·속성을 순회하며 출력
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)
````

---
##  p.11 — 하위 태그 수 확인 & type 확인

````python
# menu의 직접 하위 태그 수 (breakfast, lunch, dinner → 3)
len(root)
````

````python
# breakfast(root[0])의 item 수 (2개)
len(root[0])
````

````python
# root[0]의 타입
type(root[0])            # xml.etree.ElementTree.Element
````

---
##  p.12 — 속성 값 접근 : `get()`, `keys()`, `items()`

````python
# root[0] = Element 'breakfast'
print(root[0])                       # root의 첫 번째 자식 : Element 'breakfast'
print(root[0].get("hours"))          # get() : 'breakfast'의 'hours' 속성 값 리턴
print(root[0].keys())                # keys() : 모든 속성의 키를 리스트로 리턴
print(root[0].items())               # items() : 모든 속성의 key, value 쌍을 리턴
````

````python
# find() : 태그 이름으로 첫 번째 일치 자식 엘리먼트 반환
lunch = root.find("lunch")
print(lunch)                         # Element 'lunch'
print(lunch.get("hours"))            # 11-3
print(lunch.keys())                  # ['hours']
print(lunch.items())                 # [('hours', '11-3')]
````

````python
# lunch 타입 확인
type(lunch)                          # xml.etree.ElementTree.Element
````

---
##  p.13 — `findall()` 과 `.text` 로 태그 내용 접근

````python
# findall() : 같은 이름의 태그가 여러 개일 때 모두 리스트로 반환
print(root[0].findall('item'))       # breakfast 아래 item 엘리먼트 리스트
````

````python
# .text : 태그 안의 텍스트 내용 접근
for item in root[0].findall('item'):
    print(item.get('price'))         # 속성 값 출력
    print(item.text)                 # 태그 내용 텍스트 출력
````

````python
# 응용: 모든 식사 시간대(breakfast/lunch/dinner)의 메뉴를 정리해서 출력
for meal in root:
    print(f"\n[{meal.tag.upper()}]  운영시간: {meal.get('hours')}")
    for item in meal.findall('item'):
        print(f"  - {item.text:<25} {item.get('price')}")
````

##  p.19 — `json.loads()` : JSON 문자열 → 파이썬 객체

````python
# dumps → loads : 왕복 변환
j1 = {"name": "홍길동", "birth": "0525", "age": 30}
d1 = json.dumps(j1)          # 딕셔너리 → JSON 문자열
json.loads(d1)                # JSON 문자열 → 파이썬 딕셔너리
````

##  p.20 — JSON 파일 로드 : `json.load()`

**myinfo.json 파일 내용**
```json
{
    "name": "홍길동",
    "birth": "0525",
    "age": 30
}
```

````python
# myinfo.json 파일 생성 (실습 환경 준비)
import json

myinfo = {"name": "홍길동", "birth": "0525", "age": 30}
with open('myinfo.json', 'w', encoding='utf-8') as f:
    json.dump(myinfo, f, ensure_ascii=False, indent=4)
print("myinfo.json 생성 완료")
````

````python
# json.load() : JSON 파일을 파이썬 객체로 직접 로드
with open('./myinfo.json') as f:
    data = json.load(f)

print(type(data))            # <class 'dict'>
print(data)                  # {'name': '홍길동', 'birth': '0525', 'age': 30}
````

````python
# 실습용 page1.html 파일 생성 (교안 사이트와 동일한 HTML 구조)
html_content = """<html>
<head>
<title>A Useful Page</title>
</head>
<body>
<h1>An Interesting Title</h1>
<div>
Lorem ipsum dolor sit amet, consectetur adipisicing elit, sed do eiusmod
tempor incididunt ut labore et dolore magna aliqua. Ut enim ad minim veniam,
quis nostrud exercitation ullamco laboris nisi ut aliquip ex ea commodo
consequat. Duis aute irure dolor in reprehenderit in voluptate velit esse
cillum dolore eu fugiat nulla pariatur. Excepteur sint occaecat cupidatat
non proident, sunt in culpa qui officia deserunt mollit anim id est laborum.
</div>
</body>
</html>"""

with open('page1.html', 'w', encoding='utf-8') as f:
    f.write(html_content)
print("page1.html 생성 완료")
````

````python
# ── 교안 코드 (원본) ──────────────────────────────────────
# from urllib.request import urlopen
# html = urlopen("http://pythonscraping.com/pages/page1.html")
# print(html.read())
# ─────────────────────────────────────────────────────────

# 로컬 파일을 urlopen으로 읽기 (file:// 프로토콜 사용)
import os
from urllib.request import urlopen

local_path = 'file://' + os.path.abspath('page1.html')
html = urlopen(local_path)
content = html.read()
print(content)
````

##  p.26 — `BeautifulSoup` 으로 HTML 파싱

````python
# ── 교안 코드 (원본) ──────────────────────────────────────
# from urllib.request import urlopen
# from bs4 import BeautifulSoup
# html = urlopen("http://pythonscraping.com/pages/page1.html")
# bsObj = BeautifulSoup(html.read(), "html.parser")
# print(bsObj.h1)
# ─────────────────────────────────────────────────────────

from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen(local_path)
bsObj = BeautifulSoup(html.read(), "html.parser")

print(bsObj.h1)              # <h1>An Interesting Title</h1>
````

---
##  p.27 — BeautifulSoup 계층 탐색 & `prettify()`

````python
# 계층적 구조를 가진 태그를 다양한 경로로 직접 접근
print(bsObj.h1)              # 바로 h1 태그 접근
print(bsObj.html.body.h1)    # html → body → h1 순서로 접근
print(bsObj.body.h1)         # body → h1 접근
print(bsObj.html.h1)         # html → h1 접근
````

````python
# prettify() : 들여쓰기 포함하여 보기 좋게 HTML 구조 출력
print(bsObj.html.prettify())
````

````python
# 태그 내 텍스트만 추출 : .text 또는 .get_text()
print("제목:", bsObj.h1.text)
print("본문 일부:", bsObj.div.text.strip()[:80], "...")
````

````python
# title 태그 접근
print("title 태그:", bsObj.title)
print("title 텍스트:", bsObj.title.text)
````
