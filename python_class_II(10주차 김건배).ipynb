## 📌 p.11 — 계산기 더하기 기능 (전역변수 방식)

```python
# 전역변수를 사용하는 단일 계산기 adder 함수
result = 0

def adder(num):
    global result   # 전역변수 result 사용
    result += num
    return result

print(adder(3))   # 3
print(adder(4))   # 7
```

---
## 📌 p.12 — 만약 계산기가 두 개 필요하다면? (전역변수 2개)

```python
# 계산기가 두 개 필요한 경우 → 전역변수를 각각 따로 선언해야 하는 비효율 발생
result1 = 0
result2 = 0

def adder1(num):
    global result1
    result1 += num
    return result1

def adder2(num):
    global result2
    result2 += num
    return result2

print(adder1(3))   # 3
print(adder1(4))   # 7
print(adder2(3))   # 3
print(adder2(7))   # 10
```

## 📌 p.13 — 계산기가 여러 개 필요하다면? → 클래스 사용!

```python
# 클래스를 사용하면 계산기 객체를 여러 개 독립적으로 생성 가능
class Calculator:
    def __init__(self):
        self.result = 0       # 각 객체마다 독립적인 result 보유

    def adder(self, num):
        self.result += num
        return self.result

# 객체(인스턴스) 두 개 생성
cal1 = Calculator()
cal2 = Calculator()

# cal1 계산기 사용
print(cal1.adder(3))   # 3
print(cal1.adder(4))   # 7

# cal2 계산기 사용 (cal1과 완전히 독립)
print(cal2.adder(3))   # 3
print(cal2.adder(7))   # 10
```

## 📌 p.17 — 클래스 내부의 함수 (메서드: Method)

```python
# 메서드: 클래스 내부에 정의된 함수
# 첫 번째 파라미터 self → 메서드를 호출한 객체 자신이 자동으로 전달됨
class Service:
    secret = "지구는 4006년에 멸망한다"

    def sum(self, a, b):           # self 포함 3개의 파라미터
        result = a + b
        print("%s + %s = %s이다." % (a, b, result))

# 인스턴스 생성 후 메서드 호출
an = Service()
an.sum(1, 1)   # self는 자동 전달 → an.sum(1, 1) 호출 시 a=1, b=1 만 넘기면 됨
# 출력: 1 + 1 = 2이다.
```

## 📌 p.19 — 객체 변수 (Instance Variable)

```python
# 객체 변수: self.변수명 형태로 객체마다 고유하게 저장되는 변수
class Service:
    secret = "지구는 4006년에 멸망한다"

    def setname(self, name):
        self.name = name   # 객체 변수 생성

    def sum(self, a, b):
        result = a + b
        print("%s님, %s + %s = %s입니다." % (self.name, a, b, result))

# 인스턴스 생성 → setname 호출로 객체 변수 설정
an = Service()
an.setname("박달도사")   # an.name = "박달도사" 가 실행됨
an.sum(1, 1)
# 출력: 박달도사님, 1 + 1 = 2입니다.
```

## 📌 p.20 — 클래스 변수 vs 객체 변수 비교

```python
# 클래스 변수(secret): 모든 인스턴스가 공유
# 객체 변수(name)    : 각 인스턴스마다 독립적으로 저장

kim  = Service()
park = Service()

# 객체 변수는 각자 독립적
kim.name  = "김정보"
park.name = "박용합"

print(kim.name)    # 김정보
print(park.name)   # 박용합

# kim의 객체 변수 secret을 덮어씌워도 클래스 변수(Service.secret)는 그대로
kim.secret = "비밀은 없다"
print(park.secret)     # 지구는 4006년에 멸망한다  (클래스 변수 참조)
print(kim.secret)      # 비밀은 없다                (kim의 객체 변수로 가림)
print(Service.secret)  # 지구는 4006년에 멸망한다   (클래스 변수 원본 유지)
```

## 📌 p.21 — `__init__` 의 필요성 (setname 미호출 시 에러)

```python
# setname 없이 sum을 호출하면 AttributeError 발생
# → 객체 생성과 동시에 이름을 받도록 __init__ 이 필요!

try:
    lee = Service()   # setname 호출 안 함
    lee.sum(1, 1)     # name 속성이 없으므로 에러 발생
except AttributeError as e:
    print("AttributeError 발생:", e)
    print("→ __init__ 으로 초기화가 필요합니다!")
```

## 📌 p.22 — `__init__` 생성자로 해결

```python
# __init__: 객체 생성 시 자동 호출되는 생성자(constructor)
# → 객체를 만들면서 동시에 필요한 데이터를 초기화할 수 있음
class Service:
    secret = "지구는 4006년에 멸망한다"

    def __init__(self, name):   # 생성자: 객체 생성 시 자동 실행
        self.name = name        # 생성과 동시에 name 초기화

    def sum(self, a, b):
        result = a + b
        print("%s님, %s + %s = %s입니다." % (self.name, a, b, result))

# 인스턴스 생성 시 name을 인자로 바로 전달
an = Service("박달도사")
an.sum(1, 1)
# 출력: 박달도사님, 1 + 1 = 2입니다.
```

## 📌 p.23 — 클래스 구조 파악 (템플릿)

```python
# 클래스 전체 구조 템플릿
# ─────────────────────────────────────────
# class 클래스이름[(상속 클래스명)]:
#     <클래스 변수 1>          ← 모든 인스턴스가 공유
#     <클래스 변수 2>
#     ...
#     def 메서드1(self[, 인수1, 인수2,...]):
#         <수행할 문장 1>
#         <수행할 문장 2>
#         ...
#     def 메서드2(self[, 인수1, 인수2,...]):
#         <수행할 문장1>
#         <수행할 문장2>
#         ...
# ─────────────────────────────────────────

# 구조를 갖춘 예시 클래스
class SampleClass:
    class_var = "나는 클래스 변수 (공유)"   # 클래스 변수

    def __init__(self, value):
        self.instance_var = value           # 객체 변수 (각자 독립)

    def method1(self, x):
        return self.instance_var + x

    def method2(self):
        return self.class_var

obj = SampleClass(10)
print(obj.method1(5))   # 15
print(obj.method2())    # 나는 클래스 변수 (공유)
```

```python
# Step 2: setdata 메서드로 두 숫자를 객체 변수에 저장
class FourCal:
    def setdata(self, first, second):
        self.first  = first    # 객체 변수에 저장
        self.second = second

a = FourCal()
a.setdata(4, 2)          # first=4, second=2 저장

print(a.first)    # 4
print(a.second)   # 2
```

## 📌 p.26 — 사칙연산 클래스 FourCal (Step 3: 더하기 추가)

```python
# Step 3: sum 메서드(더하기) 추가
class FourCal:
    def setdata(self, first, second):
        self.first  = first
        self.second = second

    def sum(self):
        result = self.first + self.second
        return result

a = FourCal()
a.setdata(4, 2)

print(a.sum())   # 6
```

## 📌 p.27 — 사칙연산 클래스 FourCal (Step 4: 완성판 — +, ×, −, ÷)

```python
# Step 4: 사칙연산 메서드 4개 모두 구현 (완성된 FourCal 클래스)
class FourCal:
    def setdata(self, first, second):
        self.first  = first
        self.second = second

    def sum(self):   # 더하기
        result = self.first + self.second
        return result

    def mul(self):   # 곱하기
        result = self.first * self.second
        return result

    def sub(self):   # 빼기
        result = self.first - self.second
        return result

    def div(self):   # 나누기
        result = self.first / self.second
        return result

# ── 객체 a: first=4, second=2 ──
a = FourCal()
a.setdata(4, 2)
print("a.sum() =", a.sum())   # 6
print("a.mul() =", a.mul())   # 8
print("a.sub() =", a.sub())   # 2
print("a.div() =", a.div())   # 2.0

print()

# ── 객체 b: first=3, second=7 ──
b = FourCal()
b.setdata(3, 7)
print("b.sum() =", b.sum())   # 10
print("b.mul() =", b.mul())   # 21
print("b.sub() =", b.sub())   # -4
print("b.div() =", b.div())   # 약 0.428...
```

## 📌 p.30 — 주사위 게임 (순차적 프로그래밍 방식)

```python
# 두 명의 플레이어가 주사위 3개를 굴려 합계가 높은 사람이 이기는 게임
# ── 순차적(절차적) 프로그래밍 방식 ──
import random

player1_dice = []
player2_dice = []

for i in range(3):
    player1_dice.append(random.randint(1, 6))
    player2_dice.append(random.randint(1, 6))

print("Player 1 rolled" + str(player1_dice))
print("Player 2 rolled" + str(player2_dice))

if sum(player1_dice) == sum(player2_dice):
    print("Draw")
elif sum(player1_dice) > sum(player2_dice):
    print("Player 1 wins")
else:
    print("Player 2 wins")
```

## 📌 p.31 — 주사위 게임 (OOP 방식)

```python
# 같은 게임을 OOP(객체지향) 방식으로 구현
# → Player 클래스에 주사위(데이터)와 roll/get_dice(동작)을 캡슐화
from random import randint

class Player:
    def __init__(self):
        self.dice = []   # 주사위 결과 저장

    def roll(self):
        self.dice = []   # 이전 결과 초기화 (clears current dice)
        for i in range(3):
            self.dice.append(randint(1, 6))

    def get_dice(self):
        return self.dice

# ── 두 플레이어 객체 생성 ──
player1 = Player()
player2 = Player()

# ── 각자 주사위 굴리기 ──
player1.roll()
player2.roll()

# ── 결과 출력 ──
print("Player 1 rolled" + str(player1.get_dice()))
print("Player 2 rolled" + str(player2.get_dice()))

# ── 승패 판정 ──
if sum(player1.get_dice()) == sum(player2.get_dice()):
    print("Draw!")
elif sum(player1.get_dice()) > sum(player2.get_dice()):
    print("Player 1 wins!")
else:
    print("Player 2 wins!")
```

## 🔁 순차적 vs OOP 방식 비교 실행 (여러 번 반복 플레이)

```python
# OOP 방식의 장점: Player 객체를 재사용하여 게임을 쉽게 반복할 수 있음
print("=" * 40)
print("        🎲  주사위 게임 3라운드  🎲")
print("=" * 40)

p1 = Player()
p2 = Player()

for round_num in range(1, 4):
    p1.roll()
    p2.roll()
    p1_total = sum(p1.get_dice())
    p2_total = sum(p2.get_dice())

    print(f"\n[Round {round_num}]")
    print(f"  Player 1: {p1.get_dice()}  → 합계: {p1_total}")
    print(f"  Player 2: {p2.get_dice()}  → 합계: {p2_total}")

    if p1_total == p2_total:
        print("  → 🤝 Draw!")
    elif p1_total > p2_total:
        print("  → 🏆 Player 1 wins!")
    else:
        print("  → 🏆 Player 2 wins!")
```
