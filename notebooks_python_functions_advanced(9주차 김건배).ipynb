## 1. List Comprehension (p.4)

```python
number_list = []
number_list.append(1)
number_list.append(2)
number_list.append(3)
number_list.append(4)
number_list.append(5)
print(number_list)
```

```python
number_list = []
for number in range(1, 6):
    number_list.append(number)
print(number_list)
```

```python
number_list = list(range(1, 6))
print(number_list)
```

## 2. List Comprehension - 기본 / 함수 적용 / 조건 적용 (p.5)

```python
number_list = [number for number in range(1, 6)]
print(number_list)  # [1, 2, 3, 4, 5]
```

```python
# 함수 적용: 각 원소에 -1
number_list = [number - 1 for number in range(1, 6)]
print(number_list)  # [0, 1, 2, 3, 4]
```

```python
a_list = [number for number in range(1, 6) if number % 2 == 1]
print(a_list)  # [1, 3, 5]
```

## 3. List Comprehension - for 문과의 비교 (p.6)

```python
a_list = [number for number in range(1, 6) if number % 2 == 1]
print(a_list)
```

```python
a_list = []
for number in range(1, 6):
    if number % 2 == 1:
        a_list.append(number)
print(a_list)
```

## 4. List Comprehension 예제 (p.7)

```python
sentence = ['I', 'Love', 'Python', 'Soooooo', 'MUCH!!!']

result1 = [word.lower() for word in sentence]
print(result1)
```

```python
result2 = [word for word in sentence if len(word) > 6]
print(result2)
```

```python
result3 = [(x, x ** 2, x ** 3) for x in range(10)]
print(result3)
```

## 5. List Comprehension - 두 개의 for (p.8)

```python
rows = range(1, 4)
cols = range(1, 3)
for row in rows:
    for col in cols:
        print(row, col)
```

```python
rows = range(1, 4)
cols = range(1, 3)
cells = [(row, col) for row in rows for col in cols]
for cell in cells:
    print(cell)
```

```python
result = [(i, j) for i in range(5) for j in range(i)]
print(result)
```

## 6. Dictionary Comprehension (p.9)

```python
word = 'letters'
letter_counts = {letter: word.count(letter) for letter in word}
print(letter_counts)  # {'l': 1, 'e': 2, 't': 2, 'r': 1, 's': 1}
```

## 7. Dictionary Comprehension - 효율적인 방식 (p.10)

```python
word = 'letters'
letter_counts = {letter: word.count(letter) for letter in word}
print(letter_counts)  # {'l': 1, 'e': 2, 't': 2, 'r': 1, 's': 1}
```

```python
word = 'letters'
letter_counts = {letter: word.count(letter) for letter in set(word)}
print(letter_counts)  # {'t': 2, 'l': 1, 'r': 1, 's': 1, 'e': 2}
```

## 8. Set Comprehension (p.11)

## 9. Comprehension 추가 예제 (p.12)

```python
d = {'a': 1, 'b': 2, 'c': 3}
reversed_d = {v: k for k, v in d.items()}
print(reversed_d)
```

```python
def is_palindrome(word):
    return word == word[::-1]

hamlet = ['level', 'python', 'madam', 'code', 'racecar', 'noon']
palindromes = {word for word in hamlet if is_palindrome(word.lower())}
print(palindromes)
```

## 12. iteration - zip() (p.16)

```python
days = ['Monday', 'Tuesday', 'Wednesday']
fruits = ['banana', 'orange', 'peach']
drinks = ['coffee', 'tea', 'beer']
desserts = ['tiramisu', 'ice cream', 'pie', 'pudding']

for day, fruit, drink, dessert in zip(days, fruits, drinks, desserts):
    print(day, ": drink", drink, "- eat", fruit, "- enjoy", dessert)
```

## 13. iteration - zip() 활용: 영어-프랑스 사전 (p.17)

```python
english = 'Monday', 'Tuesday', 'Wednesday'
french = 'Lundi', 'Mardi', 'Mercredi'

print(list(zip(english, french)))
# [('Monday', 'Lundi'), ('Tuesday', 'Mardi'), ('Wednesday', 'Mercredi')]
```

```python
print(dict(zip(english, french)))
# {'Monday': 'Lundi', 'Tuesday': 'Mardi', 'Wednesday': 'Mercredi'}
```

## 14. 함수 위치 인자 (positional arguments) (p.18)

```python
# wine, entree, dessert를 받아서 딕셔너리로 만들어 반환하는 함수
def menu(wine, entree, dessert):
    return {'wine': wine, 'entree': entree, 'dessert': dessert}

print(menu('chardonnay', 'chicken', 'cake'))
```

## 17. 가능한 호출 방식 (p.21)

```python
def menu(price, wine='chardonnay', entree='chicken', dessert='pudding'):
    return {'price': price, 'wine': wine, 'entree': entree, 'dessert': dessert}

print(menu(100))                              # 1 위치 인자
print(menu(price=100))                        # 1 키워드 인자
print(menu(price=120, entree='beef'))         # 2 키워드 인자
print(menu(dessert='bagel', price=110))       # 2 키워드 인자
print(menu('eighty', 'saint-pierre', 'fish')) # 3 위치 인자
print(menu('hundred', wine='saint-pierre'))   # 1 위치 인자 + 1 키워드 인자
```

## 19. docstring (p.23)

```python
def echo(anything):
    'echo returns its input argument'
    return anything

print(echo('Hello, Python!'))
```

```python
def print_if_true(thing, check):
    """
    Prints the first argument if a second argument is true.
    The operation is:
    1. Check whether the *second* argument is true.
    2. If it is, print the *first* argument.
    """
    if check:
        print(thing)

print_if_true('Hello!', True)
print_if_true('이건 출력 안됨', False)
```
